```python
# ===============================
# Custom Sentiment Model
# ===============================

class SentimentModel:
    
    def __init__(self):
        self.positive_words = [
            "good", "great", "excellent", "happy", "love",
            "fast", "support", "appreciate", "amazing", "nice"
        ]
        
        self.negative_words = [
            "bad", "poor", "slow", "hate", "issue",
            "problem", "disappointed", "delay", "worst", "error"
        ]

    def predict(self, text, mode="normal"):
        text = text.lower()
        words = text.split()
        
        pos_count = 0
        neg_count = 0
        
        for word in words:
            if word in self.positive_words:
                pos_count += 1
            elif word in self.negative_words:
                neg_count += 1

        # Custom Modes
        if mode == "all_positive":
            return 5, "Positive"
        elif mode == "all_negative":
            return 1, "Negative"
        elif mode == "neutral_only":
            return 3, "Neutral"

        # Normal Logic
        if pos_count > neg_count:
            return 5, "Positive"
        elif neg_count > pos_count:
            return 1, "Negative"
        else:
            return 3, "Neutral"
```

```python
# ===============================
# Run Model
# ===============================

model = SentimentModel()

emails = [
    """Hello team,
    I am really disappointed with the service response time.
    The issue has not been resolved for three days.
    Please fix this urgently.""",

    """Hi,
    Great job on the new project update!
    I really appreciate your quick response and support.""",

    """Dear support,
    I have a question regarding my account.
    Can you please provide more details?"""
]

modes = ["normal", "all_positive", "all_negative", "neutral_only"]

for mode in modes:
    print("\n==============================")
    print(f"MODE: {mode.upper()}")
    print("==============================")
    
    for email in emails:
        rating, sentiment = model.predict(email, mode)
        
        print("\nEmail:\n", email)
        print("Predicted Rating:", rating, "Stars")
        print("Sentiment:", sentiment)
```
