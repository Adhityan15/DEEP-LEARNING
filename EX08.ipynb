```python
# ===============================
# Sentiment Analysis using BERT
# ===============================

# Step 1: Install required libraries (run once)
# pip install transformers torch

from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import torch.nn.functional as F

# Step 2: Load Pretrained BERT Model
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

print("Model loaded successfully!\n")

# Step 3: Function to Predict Sentiment
def predict_sentiment(text):
    
    # Tokenize input
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)
    
    # Get model output
    outputs = model(**inputs)
    logits = outputs.logits
    
    # Apply softmax
    probs = F.softmax(logits, dim=1)
    
    # Get predicted class
    predicted_class = torch.argmax(probs).item() + 1   # 1–5 stars
    
    # Map rating to sentiment
    if predicted_class <= 2:
        sentiment = "Negative"
    elif predicted_class == 3:
        sentiment = "Neutral"
    else:
        sentiment = "Positive"
    
    return predicted_class, sentiment

# Step 4: Input Emails
emails = [
    """Hello team,
    I am really disappointed with the service response time.
    The issue has not been resolved for three days.
    Please fix this urgently.""",

    """Hi,
    Great job on the new project update!
    I really appreciate your quick response and support.""",

    """Dear support,
    I have a question regarding my account.
    Can you please provide more details?"""
]

# Step 5: Display Output
for email in emails:
    rating, sentiment = predict_sentiment(email)
    
    print("====================================")
    print("Email Text:\n", email)
    print("\nPredicted Rating:", rating, "Stars")
    print("Sentiment:", sentiment)
```
