```python
import sounddevice as sd
from scipy.io.wavfile import write

fs = 44100  # Sample rate
seconds = 5  # Recording duration

print("Start speaking... 🎤")
audio = sd.rec(int(seconds * fs), samplerate=fs, channels=1)
sd.wait()

write("recorded.wav", fs, audio)

print("Recording saved as recorded.wav ✅")
```

```python
import whisper

model = whisper.load_model("base")  # first time downloads model
print("Model loaded 😌")
```

```python
model = whisper.load_model("tiny")
```

```python
result = model.transcribe("recorded.wav")
print("You said:", result["text"])
```

```python
file_path = "recorded.wav"  # change to your file if needed

result = model.transcribe(file_path)
print("Converted Text:", result["text"])
```
