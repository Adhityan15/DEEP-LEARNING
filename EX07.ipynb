```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np

text = "deep learning models are powerful and useful for ai applications."
text = text.lower()

chars = sorted(list(set(text)))
char2idx = {ch:i for i,ch in enumerate(chars)}
idx2char = {i:ch for ch,i in char2idx.items()}

vocab_size = len(chars)
seq_length = 10

inputs = []
targets = []

for i in range(len(text) - seq_length):
    seq = text[i:i+seq_length]
    target = text[i+seq_length]
    inputs.append([char2idx[c] for c in seq])
    targets.append(char2idx[target])

inputs = torch.tensor(inputs)
targets = torch.tensor(targets)

class CharLSTM(nn.Module):
    def __init__(self):
        super(CharLSTM, self).__init__()
        self.embedding = nn.Embedding(vocab_size, 64)
        self.lstm = nn.LSTM(64, 128, batch_first=True)
        self.fc = nn.Linear(128, vocab_size)
        
    def forward(self, x):
        x = self.embedding(x)
        out, _ = self.lstm(x)
        out = self.fc(out[:, -1, :])
        return out
        
model = CharLSTM()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
epochs = 50

for epoch in range(epochs):
    outputs = model(inputs)
    loss = criterion(outputs, targets)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    if (epoch+1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
print("Training Finished!")
def generate_text(seed, length=100):
    model.eval()
    result = seed
    for _ in range(length):
        seq = result[-seq_length:]
        seq_idx = torch.tensor([[char2idx[c] for c in seq]])
        with torch.no_grad():
            output = model(seq_idx)
            predicted = torch.argmax(output).item()
        result += idx2char[predicted]
    return result
    
print("\nGenerated Text:")
print(generate_text("deep learn"))
```
