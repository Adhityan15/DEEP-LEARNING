```python
# ==========================================
# Simple Chatbot using NLP (Rule-Based)
# ==========================================

# Step 1: Define Responses Dictionary
bot_responses = {
    "hello": "Hello! How can I assist you today?",
    "hi": "Hi there! What can I do for you?",
    "how are you": "I'm doing great! How about you?",
    "what is your name": "I'm your friendly chatbot 😊",
    "recommend a book": "Sure, I recommend 'The Alchemist' by Paulo Coelho. It's a great read!",
    "bye": "Goodbye! Have a great day!"
}

# Step 2: Response Generation Function
def generate_response(user_input):
    
    # Convert input to lowercase
    user_input = user_input.lower()
    
    # Check for match
    if user_input in bot_responses:
        return bot_responses[user_input]
    else:
        return "Sorry, I didn't understand that. Can you try asking differently?"

# Step 3: Chatbot Conversation Loop
def chatbot():
    
    print("Chatbot: Hi there! I'm your friendly chatbot. Feel free to start a conversation or type 'bye' to end.")
    
    while True:
        user_input = input("You: ")
        
        # Exit condition
        if user_input.lower() == "bye":
            print("Chatbot:", bot_responses["bye"])
            break
        
        # Generate response
        response = generate_response(user_input)
        print("Chatbot:", response)

# Step 4: Run Chatbot
chatbot()
```
